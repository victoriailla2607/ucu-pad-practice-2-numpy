# Práctica de NumPy

## Análisis de distribuciones probabilísticas con NumPy y Matplotlib

En esta práctica se generan dos muestras de 1.000 observaciones: una distribución normal y una uniforme. Luego se calculan sus estadísticos descriptivos y se comparan mediante histogramas.


```python
import matplotlib.pyplot as plt
import numpy as np

# Parámetros y generación reproducible de las muestras.
rng = np.random.default_rng(42)
n_observaciones = 1_000

media_normal = 10
desvio_normal = 2
limite_inferior = 0
limite_superior = 20

distribucion_normal = rng.normal(
    loc=media_normal,
    scale=desvio_normal,
    size=n_observaciones,
)
distribucion_uniforme = rng.uniform(
    low=limite_inferior,
    high=limite_superior,
    size=n_observaciones,
)
```

## Estadísticos descriptivos

Se calculan la media, mediana, varianza, desvío estándar, mínimo y máximo de cada muestra. Se usa ddof=0, por lo que la varianza y el desvío corresponden a los de la población generada.


```python
def calcular_estadisticos(datos):
    return {
        "Media": np.mean(datos),
        "Mediana": np.median(datos),
        "Varianza": np.var(datos),
        "Desvío estándar": np.std(datos),
        "Mínimo": np.min(datos),
        "Máximo": np.max(datos),
    }


estadisticos = {
    "Normal": calcular_estadisticos(distribucion_normal),
    "Uniforme": calcular_estadisticos(distribucion_uniforme),
}

for nombre, valores in estadisticos.items():
    print(f"Distribución {nombre}:")
    for estadistico, valor in valores.items():
        print(f"  {estadistico}: {valor:.2f}")
    print()
```

## Histogramas

Los dos histogramas se muestran en una misma figura para facilitar la comparación.


```python
fig, ejes = plt.subplots(1, 2, figsize=(12, 4), sharey=True)

ejes[0].hist(
    distribucion_normal,
    bins=30,
    color="steelblue",
    edgecolor="white",
    alpha=0.85,
    label="Muestra normal",
)
ejes[0].set_title("Distribución normal")
ejes[0].set_xlabel("Valor")
ejes[0].set_ylabel("Frecuencia")
ejes[0].legend()
ejes[0].grid(axis="y", alpha=0.3)

ejes[1].hist(
    distribucion_uniforme,
    bins=30,
    color="darkorange",
    edgecolor="white",
    alpha=0.85,
    label="Muestra uniforme",
)
ejes[1].set_title("Distribución uniforme")
ejes[1].set_xlabel("Valor")
ejes[1].set_ylabel("Frecuencia")
ejes[1].legend()
ejes[1].grid(axis="y", alpha=0.3)

fig.suptitle("Comparación de distribuciones aleatorias")
fig.tight_layout()
plt.show()
```

## Reflexión

La muestra normal se concentra alrededor de su media (10) y su histograma tiene una forma aproximadamente acampanada y simétrica; la mayoría de sus valores queda cerca del centro y hay menos observaciones en los extremos. La muestra uniforme, en cambio, reparte sus observaciones de forma aproximadamente pareja entre 0 y 20, por lo que su histograma tiende a ser más plano.

Ambas distribuciones tienen una media teórica cercana a 10, aunque las medias y medianas de las muestras pueden diferir ligeramente por azar. La uniforme cubre un intervalo mucho más amplio: su varianza teórica es (20 - 0)^2 / 12, aproximadamente 33,33, frente a la varianza 4 de la normal elegida. Por eso, sus valores están más dispersos respecto del centro. La semilla fija permite obtener los mismos resultados cada vez que se ejecuta el notebook.
