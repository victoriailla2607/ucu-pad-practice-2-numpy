# Práctica de NumPy

## Comparación de una distribución normal y una uniforme

En este notebook vamos a crear dos grupos de números al azar, calcular algunos valores para describirlos y dibujar sus histogramas.


## 1. Importar las bibliotecas

NumPy nos ayuda a trabajar con números y Matplotlib nos permite hacer gráficos.


```python
import numpy as np
import matplotlib.pyplot as plt
```

## 2. Crear las distribuciones

Vamos a crear 1.000 números para cada distribución. La normal tendrá una media de 10 y un desvío estándar de 2. La uniforme generará números entre 0 y 20. Usamos una semilla para que los números sean los mismos cada vez que ejecutamos el notebook.


```python
np.random.seed(42)

cantidad = 1000

# Crear los números de la distribución normal.
datos_normales = np.random.normal(
    loc=10,       # Media
    scale=2,      # Desvío estándar
    size=cantidad # Cantidad de números
)

# Crear los números de la distribución uniforme.
datos_uniformes = np.random.uniform(
    low=0,        # Valor mínimo del rango
    high=20,      # Valor máximo del rango
    size=cantidad # Cantidad de números
)
```

## 3. Calcular los estadísticos

Calculamos la media, la mediana, la varianza, el desvío estándar, el mínimo y el máximo para cada grupo de números.


```python
# Estadísticos de la distribución normal.
print("Distribución normal")
print("Media:", np.mean(datos_normales))
print("Mediana:", np.median(datos_normales))
print("Varianza:", np.var(datos_normales))
print("Desvío estándar:", np.std(datos_normales))
print("Mínimo:", np.min(datos_normales))
print("Máximo:", np.max(datos_normales))

# Estadísticos de la distribución uniforme.
print("\nDistribución uniforme")
print("Media:", np.mean(datos_uniformes))
print("Mediana:", np.median(datos_uniformes))
print("Varianza:", np.var(datos_uniformes))
print("Desvío estándar:", np.std(datos_uniformes))
print("Mínimo:", np.min(datos_uniformes))
print("Máximo:", np.max(datos_uniformes))
```

## 4. Dibujar los histogramas

Cada histograma muestra cuántos números caen dentro de distintos intervalos. Los dos gráficos aparecen juntos para compararlos.


```python
# Crear una figura con dos gráficos, uno al lado del otro.
fig, ejes = plt.subplots(1, 2, figsize=(12, 4))

# Histograma de la distribución normal.
ejes[0].hist(datos_normales, bins=30, color="steelblue", label="Normal")
ejes[0].set_title("Distribución normal")
ejes[0].set_xlabel("Valor")
ejes[0].set_ylabel("Cantidad de números")
ejes[0].legend()
ejes[0].grid(True, alpha=0.3)

# Histograma de la distribución uniforme.
ejes[1].hist(datos_uniformes, bins=30, color="darkorange", label="Uniforme")
ejes[1].set_title("Distribución uniforme")
ejes[1].set_xlabel("Valor")
ejes[1].set_ylabel("Cantidad de números")
ejes[1].legend()
ejes[1].grid(True, alpha=0.3)

fig.suptitle("Comparación de las dos distribuciones")
plt.tight_layout()
plt.show()
```

## 5. Reflexión

En el histograma normal, la mayoría de los números se encuentra cerca de 10. Por eso, el gráfico suele tener una forma parecida a una campana. En el histograma uniforme, los números están repartidos de manera más pareja entre 0 y 20.

Las dos distribuciones tienen una media cercana a 10. Sin embargo, la distribución uniforme tiene más dispersión: sus números pueden estar en casi todo el rango de 0 a 20, mientras que los números normales suelen estar más cerca de 10. Los resultados exactos de los estadísticos se muestran al ejecutar la celda anterior.
